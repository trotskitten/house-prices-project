## 3. Data Cleaning and Feature Engineering

**What this cleaning step does:**
- `date` is parsed and split into `sale_year` and `sale_month`, then dropped.
- The single row with 33 bedrooms is removed as a clear data-entry error.
- `id` and `zipcode` are **kept** in `df_model` — `id` will be excluded when building the feature matrix X, and decisions on how to handle `zipcode` are deferred to the modelling notebooks.

```python
df_model = df.copy()

# Parse date and add useful date-based features
df_model["date"] = pd.to_datetime(df_model["date"], format="%Y%m%dT%H%M%S")
df_model["sale_year"] = df_model["date"].dt.year
df_model["sale_month"] = df_model["date"].dt.month

df_model = df_model[df_model['bedrooms'] < 33]

# Convert selected columns to categorical/string for consistent encoding
# df_model["zipcode"] = df_model["zipcode"].astype(str)

# Keep id only for reference, do not use as predictor
# df_model = df_model.drop(columns=["id", "date",'zipcode'])
df_model = df_model.drop(columns=["date"])

print(df_model.shape)
df_model.head()
df_model["bedrooms"].describe()
```

**Data Cleaning Decisions**

Two deliberate choices are made here:

1. **Date parsing → `sale_year` + `sale_month`**: The raw `date` string carries temporal information the model can use (e.g., seasonality in housing markets). Splitting it into two numeric components lets models use the signal without needing any further transformation. The original `date` column is then dropped.

2. **Bedroom outlier removal**: The single record with 33 bedrooms is removed — it is almost certainly a data-entry error and would distort model training.

`id` and `zipcode` are **retained** in `df_model`. `id` will be dropped when constructing the feature matrix X. Decisions on how to handle `zipcode` (keep numeric, encode categorically, or drop in favour of lat/long) are made separately in each modelling notebook.

**Price Distribution Interpretation**

The histogram reveals a strong **right-skew**: the bulk of sales cluster between $200K–$700K, but a long tail extends to $7.7M. The KDE curve reinforces this — the peak (mode) is well below the mean. The boxplot makes the outlier situation concrete: the whiskers extend to roughly $1.1M, and every point beyond that is flagged as an outlier, representing genuine luxury sales, not data errors.

**Implication for modeling**: right-skewed targets can inflate RMSE because large errors on high-value homes are squared. A log-transform on `price` could normalize the distribution and produce better-calibrated predictions, though it adds complexity in interpreting the output. For this project we model on the raw scale and accept that RMSE will be dominated by the upper tail.

## 6. Prepare Features for Modeling

```python
target_col = "price"
feature_cols = [c for c in df_model.columns if c != target_col]

X = df_model[feature_cols].copy()
X = X.drop('id', axis = 1)
y = df_model[target_col].copy()

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)


print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
X.head(10)
```

**Feature Preparation Interpretation**

The feature matrix **X** contains **20 columns**: all `df_model` columns except `price` (target) and `id` (excluded as a non-predictive identifier). `zipcode` is retained as a raw integer at this stage — no encoding is applied here.

The **80/20 train-test split** with `random_state=42` ensures:
- **80% (17,289 rows)** are used for fitting the models.
- **20% (4,323 rows)** are held out and never seen during training — this is our unbiased estimate of how the model generalises to new homes.
- Fixing the seed makes all results reproducible across re-runs.
